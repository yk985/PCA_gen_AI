```python
Jtens_PCA = load_JPCA_tensor(model=3)
J = Jtens_PCA
```

```python
import torch
import matplotlib.pyplot as plt
import numpy as np

# Assuming your tensor is named J
# J shape: [21, 1225, 63, 1]

# Remove the last singleton dimension for easier processing
J = J.squeeze(-1)  # shape now [21, 1225, 63]

# 1. Basic statistics: mean, std, min, max along different axes
print("Overall stats:")
print(f"Mean: {J.mean().item():.4f}")
print(f"Std: {J.std().item():.4f}")
print(f"Min: {J.min().item():.4f}")
print(f"Max: {J.max().item():.4f}")

# Mean and std per PCA component (axis=0)
mean_per_pca = J.mean(dim=(1,2))  # shape [21]
std_per_pca = J.std(dim=(1,2))    # shape [21]
print("Mean per PCA component:", mean_per_pca)
print("Std per PCA component:", std_per_pca)

# 2. Correlation matrix between PCA components across all amino acids and positions
# Flatten last two dims
J_flat = J.reshape(21, -1)

# Convert to numpy for correlation
J_np = J_flat.cpu().numpy()
corr = np.corrcoef(J_np)  # shape [21, 21]

plt.figure(figsize=(8,6))
plt.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(label='Correlation')
plt.title('Correlation between PCA components')
plt.xlabel('PCA component')
plt.ylabel('PCA component')
plt.show()

# 3. Histogram of values to check distribution
plt.figure(figsize=(6,4))
plt.hist(J.flatten().cpu().numpy(), bins=50)
plt.title("Distribution of tensor values")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

# 4. Visualize one PCA component across amino acid positions (e.g. PCA 0)
pca_idx = 0
# Shape: [1225, 63]
slice_pca = J[pca_idx].cpu().numpy()

plt.figure(figsize=(10,6))
plt.imshow(slice_pca, aspect='auto', cmap='viridis')
plt.colorbar(label='Value')
plt.title(f'PCA component {pca_idx} values across positions and amino acids')
plt.xlabel('Amino acid index (0-62)')
plt.ylabel('Position / pair index (0-1224)')
plt.show()
```
