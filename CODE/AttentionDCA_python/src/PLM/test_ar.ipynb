```python
def energy_position_k_loop_oneseq(k, aa, seq, J):
    print(seq.shape)
    e_vec = 0
    # same but vectorized
    for j in range(k):
        # J[k, j, aa, seq[j]] for all j<k
        e_vec += J[k, j, aa, seq[j]]
    #e_vec += np.sum(J[k, :k, aa, seq[:k]])
    return e_vec

def energy_position_k_loop(k, aa, seq_array, J):
    e = np.zeros(seq_array.shape[0])
    for i in range(seq_array.shape[0]):
        e[i] = energy_position_k_loop_oneseq(k, aa, seq_array[i], J)
    return e
```

```python
def energy_position_k_all(k, aa, seq_array, J):
    n_seq = seq_array.shape[0]
    # indices of positions < k
    idx = np.arange(k)

    # shape (n_seq, k)
    seq_sub = seq_array[:, idx]
    print(seq_sub.shape)

    # gather J[k, j, aa, seq_sub] for all j<k, seqs
    # this broadcasts: (k,) vs (n_seq, k)
    e = J[k, idx[None, :], aa, seq_sub]
    print(e.shape)

    # sum over positions <k, leaving (n_seq,)
    return e.sum(axis=1)
```
